# hinge loss and gradient

```python
# formula: max(0, 1 - y*f(x))

def hinge_loss(X, y, w, b):
    n = len(X)
    total_loss = 0.0
    for i in range(n):
        margin = y[i] * (dot(w, X[i]) + b) # y_i * (w^T x_i + b)
        total_loss += max(0.0, 1.0 - margin) # sum(max(0, 1 - margin))
    return total_loss / n # average hinge loss
```

# linear SVM via grad desc

```python
class LinearSVM:
    def __init__(self, lr=0.001, lambda_param=0.01, n_epochs=1000):
        self.lr = lr
        self.lambda_param = lambda_param
        self.n_epochs = n_epochs
        self.w = None
        self.b = 0.0

    def fit(self, X, y):
        n_features = len(X[0])
        self.w = [0.0] * n_features
        self.b = 0.0

        for epoch in range(self.n_epochs):
            # for each sample in X
            for i in range(len(X)):
                # margin determines if its inside or outside the support vector planes
                margin = y[i] * (dot(self.w, X[i]) + self.b) # y_i * (w^T x_i + b)
                if margin >= 1: # point is either on the margin (is support vec) or classified correctly
                     # nudge in opposite direction of gradient lambda * w, scaled by learning rate
                    self.w = [wj - self.lr * self.lambda_param * wj
                              for wj in self.w]
                    # b does not change in this case
                else: # this occurs when the point is either inside the margin or misclassified
                    # nudge in opposite direction of gradient lambda * w - y_i * x_i, scaled by learning rate
                    self.w = [wj - self.lr * (self.lambda_param * wj - y[i] * X[i][j])
                              for j, wj in enumerate(self.w)]
                    # nudge in opposite direction of gradient -y_i, scaled by learning rate
                    self.b -= self.lr * (-y[i])

    def predict(self, X):
        # classify based on which side of boundary the data falls on
        return [1 if dot(self.w, x) + self.b >= 0 else -1 for x in X]
```

# kernel functions
linear, polynomial and RBF kernels. brings data up to higher dimensional space so that it is separable.

```python
# K(x, z) = x . z
def linear_kernel(x, z):
    return dot(x, z)

# K(x, z) = (x . z + c)^d
def polynomial_kernel(x, z, degree=3, c=1.0):
    return (dot(x, z) + c) ** degree

# K(x, z) = exp(-gamma * ||x - z||^2)
def rbf_kernel(x, z, gamma=0.5):
    diff = [xi - zi for xi, zi in zip(x, z)]
    return math.exp(-gamma * dot(diff, diff))
```

# margin and support vector identification
identify support vectors and compute margin width.


```python
def find_support_vectors(X, y, w, b, tol=1e-3):
    support_vectors = []

    for i in range(len(X)): # for all samples
        # compute the margin in the usual way
        margin = y[i] * (dot(w, X[i]) + b)

        # add point to support vec list if the margin - 1 is close to 0 (within tol)
        if abs(margin - 1.0) < tol:
            support_vectors.append(i)
    return support_vectors
```
